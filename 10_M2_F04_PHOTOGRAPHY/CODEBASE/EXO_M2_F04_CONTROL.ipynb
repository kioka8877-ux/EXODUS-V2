```python
# ── Configuration ──────────────────────────────────────────────────────────
DRIVE_ROOT   = '/content/drive/MyDrive/DRIVE_EXODUS_V2'
FREGATE_DIR  = f'{DRIVE_ROOT}/10_M2_F04_PHOTOGRAPHY'

# Optionnel : chemin explicite vers le .blend de M2_F03
SCENE_BLEND  = ''  # Laisser vide pour auto-détection dans IN_SCENE_BLEND/

BLENDER_PATH = ''  # Laisser vide pour auto-détection
PRESET       = 'production'  # 'production' ou 'preview'
NO_DOF       = False
NO_ATMO      = False
SHAKE_PRESET = 'handheld'  # 'handheld', 'subtle', 'aggressive'
VERBOSE      = True
DRY_RUN      = False

print('✓ Configuration chargée')
```

```python
# ── Mount Drive ────────────────────────────────────────────────────────────
from google.colab import drive
drive.mount('/content/drive')
print('✓ Drive monté')
```

```python
# ── Vérification des inputs ────────────────────────────────────────────────
import os
from pathlib import Path

in_scene_dir = Path(FREGATE_DIR) / 'IN_SCENE_BLEND'
blends = list(in_scene_dir.glob('*.blend'))

print(f'IN_SCENE_BLEND/ — {len(blends)} fichier(s):')
for b in blends:
    print(f'  → {b.name}  ({b.stat().st_size // 1024} KB)')

if not blends:
    print('⚠ Aucun .blend trouvé — placez le scene_ready.blend de M2_F03 dans IN_SCENE_BLEND/')
```

```python
# ── Lancement M2_F04 ───────────────────────────────────────────────────────
import subprocess, sys

codebase = f'{FREGATE_DIR}/CODEBASE'
cmd = [
    sys.executable,
    f'{codebase}/EXO_M2_F04_PHOTOGRAPHY.py',
    '--drive-root', DRIVE_ROOT,
    '--preset', PRESET,
    '--shake-preset', SHAKE_PRESET,
]

if SCENE_BLEND:  cmd += ['--scene', SCENE_BLEND]
if BLENDER_PATH: cmd += ['--blender-path', BLENDER_PATH]
if NO_DOF:       cmd.append('--no-dof')
if NO_ATMO:      cmd.append('--no-atmosphere')
if VERBOSE:      cmd.append('--verbose')
if DRY_RUN:      cmd.append('--dry-run')

print('Lancement M2_F04...')
result = subprocess.run(cmd, capture_output=False, text=True)
print(f'\nCode retour: {result.returncode}')
```

```python
# ── Lecture du rapport ─────────────────────────────────────────────────────
import json
from pathlib import Path

report_path = Path(FREGATE_DIR) / 'OUT_REPORT' / 'm2_f04_report.json'
if report_path.exists():
    with open(report_path) as f:
        report = json.load(f)
    print(f"Statut  : {report['status']}")
    print(f"Version : {report['version']}")
    if report.get('outputs', {}).get('scene_with_camera'):
        print(f"Output  : {report['outputs']['scene_with_camera']}")
else:
    print('Rapport non trouvé — lancer la cellule précédente d\'abord')
```
