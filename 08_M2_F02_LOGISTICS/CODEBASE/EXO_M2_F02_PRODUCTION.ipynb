````python
# ── CELLULE 0 — CONFIGURATION ────────────────────────────────
import subprocess
from pathlib import Path

FREGATE_ROOT = Path("../")
CODEBASE     = FREGATE_ROOT / "CODEBASE"
SCRIPT       = CODEBASE / "EXO_M2_F02_LOGISTICS.py"

# ── Paramètres ────────────────────────────────────────────────
BLENDER_PATH = None       # None = auto-détection (Drive EXODUS ou PATH système)
GLB_AVATAR   = None       # None = auto-détection dans IN_GLB_AVATAR/
OUTPUT_NAME  = "actor_equipped"
BYPASS       = False      # True = skip props, copie directe
VERBOSE      = True

print(f"Script    : {SCRIPT}")
print(f"Bypass    : {BYPASS}")
print(f"Output    : {OUTPUT_NAME}")
````

````python
# ── CELLULE 1 — DRY-RUN (validation sans exécution) ──────────
cmd = ["python", str(SCRIPT), "--dry-run", "--verbose"]
if GLB_AVATAR:
    cmd += ["--glb", GLB_AVATAR]
if BYPASS:
    cmd += ["--bypass"]

result = subprocess.run(cmd, capture_output=True, text=True, cwd=str(CODEBASE))
print(result.stdout)
if result.returncode != 0:
    print("STDERR:", result.stderr)
````

````python
# ── CELLULE 2 — EXÉCUTION COMPLÈTE ───────────────────────────
cmd = ["python", str(SCRIPT), "--output-name", OUTPUT_NAME]
if GLB_AVATAR:
    cmd += ["--glb", GLB_AVATAR]
if BLENDER_PATH:
    cmd += ["--blender-path", BLENDER_PATH]
if BYPASS:
    cmd += ["--bypass"]
if VERBOSE:
    cmd += ["--verbose"]

print("Commande:", " ".join(cmd))
result = subprocess.run(cmd, capture_output=True, text=True, cwd=str(CODEBASE))
print(result.stdout)
if result.returncode != 0:
    print("ERREUR:", result.stderr)
````

````python
# ── CELLULE 3 — LECTURE RAPPORT ──────────────────────────────
import json
report_path = FREGATE_ROOT / "OUT_REPORT" / "m2_f02_report.json"
if report_path.exists():
    with open(report_path) as f:
        report = json.load(f)
    print(f"Statut    : {report.get('status')}")
    print(f"Mode      : {report.get('mode', 'NORMAL')}")
    print(f"Pipeline  : {report.get('pipeline')}")
    inp = report.get('input', {})
    print(f"Props     : {inp.get('props_resolved', '-')}/{inp.get('props_resolved', 0) + len(inp.get('props_missing', []))}")
    print(f"Suivant   : {report.get('next_fregate')}")
else:
    print("Rapport non trouvé — frégate non exécutée")
````
