**Rôle :** Vérification pré-vol et diagnostics de la frégate M2_F02.

| R-03 | Audio prime sur animation GLB |

**Flux :** `M2_F01/OUT_VALIDATED/` → **M2_F02** → `M2_F03/IN_GLB_AVATAR/`

```python
# ── CELLULE 0 — CONFIGURATION ────────────────────────────────
from pathlib import Path
import json

FREGATE_ROOT  = Path("../")  # 08_M2_F02_LOGISTICS/
CODEBASE      = FREGATE_ROOT / "CODEBASE"
IN_GLB        = FREGATE_ROOT / "IN_GLB_AVATAR"
IN_PROPS      = FREGATE_ROOT / "IN_PROPS_LIBRARY"
IN_PLAN       = FREGATE_ROOT / "IN_PRODUCTION_PLAN"
OUT_BAKED     = FREGATE_ROOT / "OUT_BAKED_ACTORS"
OUT_REPORT    = FREGATE_ROOT / "OUT_REPORT"

print(f"Frégate  : 08_M2_F02_LOGISTICS")
print(f"Pipeline : MODE 2 — FROM SCRATCH")
print(f"Racine   : {FREGATE_ROOT.resolve()}")
```

```python
# ── CELLULE 1 — VÉRIFICATION STRUCTURE ───────────────────────
dirs = [IN_GLB, IN_PROPS, IN_PLAN, OUT_BAKED, OUT_REPORT]
for d in dirs:
    status = "✅" if d.exists() else "❌"
    print(f"  {status} {d.name}")

scripts = [
    "EXO_M2_F02_LOGISTICS.py",
    "socketing_engine.py",
    "actor_assembly.py",
    "props_loader.py",
    "final_baker.py",
    "timeline_manager.py",
    "requirements.txt",
]
print("\nScripts:")
for s in scripts:
    p = CODEBASE / s
    print(f"  {'✅' if p.exists() else '❌'} {s}")
```

```python
# ── CELLULE 2 — INSPECTION GLB AVATAR ────────────────────────
glbs = list(IN_GLB.glob("*.glb"))
print(f"GLB avatars trouvés : {len(glbs)}")
for g in glbs:
    size_mb = g.stat().st_size / (1024 * 1024)
    print(f"  • {g.name} — {size_mb:.2f} MB")

if not glbs:
    print("  ⚠️  Aucun GLB — déposez avatar_validated.glb depuis M2_F01/OUT_VALIDATED/")
```

```python
# ── CELLULE 3 — INSPECTION PROPS LIBRARY ─────────────────────
supported = [".glb", ".gltf", ".fbx", ".blend", ".obj"]
props = [f for ext in supported for f in IN_PROPS.glob(f"*{ext}")]
print(f"Props disponibles : {len(props)}")
for p in props:
    size_kb = p.stat().st_size / 1024
    print(f"  • {p.name} — {size_kb:.1f} KB")

if not props:
    print("  ℹ️  Aucun prop — bypass automatique activé")
```

```python
# ── CELLULE 4 — PRODUCTION PLAN ──────────────────────────────
plans = list(IN_PLAN.glob("*.json"))
print(f"Plans trouvés : {len(plans)}")
for plan_path in plans:
    with open(plan_path) as f:
        plan = json.load(f)
    scenes = plan.get("scenes", [])
    total_props = sum(len(s.get("props_actions", [])) for s in scenes)
    req_u02 = plan.get("production_notes", {}).get("requires_u02", True)
    print(f"  • {plan_path.name}")
    print(f"    Scènes: {len(scenes)} | Props actions: {total_props} | requires_u02: {req_u02}")

if not plans:
    print("  ℹ️  Aucun plan → bypass automatique")
```

```python
# ── CELLULE 5 — OUTPUTS ───────────────────────────────────────
print("Outputs générés :")
for d, exts in [(OUT_BAKED, [".abc", ".blend", ".glb"]), (OUT_REPORT, [".json"])]:
    files = [f for ext in exts for f in d.glob(f"*{ext}")]
    print(f"  {d.name}/")
    for f in files:
        size_mb = f.stat().st_size / (1024 * 1024)
        print(f"    • {f.name} — {size_mb:.2f} MB")
    if not files:
        print(f"    (vide)")
```
