### Mode d'emploi

4. **Cellule 4** — Routing final + instructions d'exécution

````python
#@title 🔗 CELLULE 1 — Montage Drive + Configuration
#@markdown Le projet est installé directement sur Drive (persistant entre sessions).

from google.colab import drive
drive.mount('/content/drive')

import os, sys, subprocess
from pathlib import Path

# Racine du projet sur Drive — persistante
DRIVE_ROOT = "/content/drive/MyDrive/EXODUS_V2"  #@param {type:"string"}
REPO_LOCAL = DRIVE_ROOT  # Le repo EST sur Drive

# Cloner dans Drive si absent
if not Path(REPO_LOCAL).exists():
    print("⏳ Premier lancement — clonage dans Drive (peut prendre 1-2 min)...")
    subprocess.run(
        ["git", "clone", "--depth", "1",
         "https://github.com/kioka8877-ux/EXODUS-V2.git",
         REPO_LOCAL],
        check=True
    )
    print("✓ Repo cloné sur Drive — persistant pour toutes les sessions")
else:
    print("✓ Projet déjà présent sur Drive")
    # Pull silencieux pour rester à jour
    result = subprocess.run(
        ["git", "-C", REPO_LOCAL, "pull", "--ff-only"],
        capture_output=True, text=True
    )
    if "Already up to date" in result.stdout:
        print("✓ Code à jour")
    else:
        print(f"✓ Mis à jour : {result.stdout.strip()}")

sys.path.insert(0, REPO_LOCAL)

print(f"✓ Projet Drive : {REPO_LOCAL}")
````

````python
#@title 🎯 CELLULE 2 — Choix du Mode Pipeline
#@markdown Sélectionnez le mode de production.

MODE = 2  #@param [1, 2] {type:"integer"}
#@markdown - **1** = VIDEO-TO-VIDEO (Pipeline Sacré — vidéo virale en entrée)
#@markdown - **2** = FROM SCRATCH (Pipeline Forgé — avatar GLB animé en entrée)

MODE_LABELS = {
    1: "VIDEO-TO-VIDEO — Pipeline Sacré",
    2: "FROM SCRATCH — Pipeline Forgé",
}

MODE_UNITS = {
    1: {
        "U00": "00_CORTEX_HQ/CODEBASE/EXO_00_CORTEX.py",
        "U01": "01_ANIMATION_ENGINE/CODEBASE/EXO_01_TRANSMUTATION.py",
        "U02": "02_LOGISTICS_DEPOT/CODEBASE/EXO_02_LOGISTICS.py",
        "U03": "03_SCENOGRAPHY_DOCK/CODEBASE/EXO_03_SCENOGRAPHY.py",
        "U04": "04_PHOTOGRAPHY_WING/CODEBASE/EXO_04_PHOTOGRAPHY.py",
        "U05": "05_ALCHEMIST_LAB/CODEBASE/EXO_05_ALCHEMIST.py",
        "U06": "06_AIRCRAFT_CARRIER/CODEBASE/EXO_06_CARRIER.py",
    },
    2: {
        "F01": "07_M2_F01_ANIMATION/CODEBASE/EXO_M2_F01_ANIMATION.py",
        "F02": "08_M2_F02_LOGISTICS/CODEBASE/EXO_M2_F02_LOGISTICS.py",
        "F03": "09_M2_F03_SCENOGRAPHY/CODEBASE/EXO_M2_F03_SCENOGRAPHY.py",
        "F04": "10_M2_F04_PHOTOGRAPHY/CODEBASE/EXO_M2_F04_PHOTOGRAPHY.py",
        "F05": "11_M2_F05_ALCHEMIST/CODEBASE/EXO_M2_F05_ALCHEMIST.py",
        "F06": "12_M2_F06_CARRIER/CODEBASE/EXO_M2_F06_CARRIER.py",
    },
}

print(f"✓ Mode sélectionné : {MODE} — {MODE_LABELS[MODE]}")
````

````python
#@title 🔍 CELLULE 3 — Routing Dry-Run (vérification)
#@markdown Affiche le routing et vérifie la présence des frégates.

from pathlib import Path

print(f"\n{'='*60}")
print(f"  ROUTING — MODE {MODE} : {MODE_LABELS[MODE]}")
print(f"  Repo Local : {REPO_LOCAL}")
print(f"{'='*60}")

units = MODE_UNITS[MODE]
all_ok = True

print("\n  Frégates :")
for unit_id, rel_path in units.items():
    abs_path = Path(REPO_LOCAL) / rel_path
    exists = abs_path.exists()
    status = "✓" if exists else "✗ ABSENT"
    if not exists:
        all_ok = False
    print(f"    [{unit_id}] {status} — {rel_path}")

print()
if all_ok:
    print("  ✓ Toutes les frégates présentes. Routing opérationnel.")
else:
    print("  ⚠ Certaines frégates sont absentes.")
    print("    → Re-executer C1 pour recloner le repo.")

print()
print("  [DRY-RUN] Aucune exécution. Vérification uniquement.")
````

````python
#@title ▶️ CELLULE 4 — Routing Final + Instructions
#@markdown Affiche les commandes à exécuter dans l'ordre.

EXTRA_ARGS = ""  #@param {type:"string"}
#@markdown Arguments optionnels à passer aux orchestrateurs (ex: --verbose, --dry-run)

print(f"\n{'='*60}")
print(f"  EXODUS V2 — MODE {MODE} : {MODE_LABELS[MODE]}")
print(f"  LOI II : L'Empereur transfère les fichiers entre frégates.")
print(f"{'='*60}")

if MODE == 1:
    print("\n  INPUT REQUIS :")
    print("    00_CORTEX_HQ/IN_VIDEO_SOURCE/ → vidéo_source.mp4")
elif MODE == 2:
    print("\n  INPUTS REQUIS :")
    print("    M2_01_ANIMATION_ENGINE/IN_AVATAR_GLB/ → avatar.glb (avec animations)")
    print("    M2_01_ANIMATION_ENGINE/IN_AUDIO/      → audio.wav (optionnel)")
    print("    M2_03_SCENOGRAPHY_DOCK/IN_DECOR_GLB/  → decor.glb (fourni Opérateur)")

print("\n  ORDRE D'EXÉCUTION :")
units = MODE_UNITS[MODE]
for i, (unit_id, rel_path) in enumerate(units.items(), 1):
    abs_path = Path(REPO_LOCAL) / rel_path
    cmd = f"python {abs_path}"
    if EXTRA_ARGS:
        cmd += f" {EXTRA_ARGS}"
    print(f"\n  ÉTAPE {i} [{unit_id}]:")
    print(f"    {cmd}")

print()
print("  ► ROUTING TERMINÉ. Exécution manuelle requise.")
print("  AD MAJOREM GLORIAM IMPERATORIS.")
````
